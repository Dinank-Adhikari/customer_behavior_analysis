```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ' , '_')
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# Create column Purcahse_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
# Step 1: Inspect the three similar columns to see which one has the real values
cols = ['Purcahse_frequency_days', 'Purchase_frequency_days', 'purchase_frequency_days']
print(df[cols].head())
```

```python
# Step 2: Drop the two duplicate columns (run only once)
df = df.drop(columns=['Purcahse_frequency_days', 'Purchase_frequency_days'])
```

```python
# Step 3: Standardize column names to lowercase and check for case-duplicates (should print [])
df.columns = df.columns.str.strip().str.lower()
print(df.columns[df.columns.str.lower().duplicated(keep=False)].tolist())
```

```python
# Step 4: Convert age_group to text and write the cleaned DataFrame to SQL Server as dbo.customer
from sqlalchemy.types import VARCHAR

df['age_group'] = df['age_group'].astype(str)

df.to_sql(
    "customer", engine, schema="dbo", if_exists="replace", index=False,
    dtype={
        "gender": VARCHAR(20),
        "item_purchased": VARCHAR(50),
        "category": VARCHAR(50),
        "location": VARCHAR(50),
        "size": VARCHAR(10),
        "color": VARCHAR(30),
        "season": VARCHAR(20),
        "subscription_status": VARCHAR(10),
        "shipping_type": VARCHAR(30),
        "discount_applied": VARCHAR(10),
        "payment_method": VARCHAR(30),
        "frequency_of_purchases": VARCHAR(30),
        "age_group": VARCHAR(20),
    },
)
```

```python
# Step 5: Verify the table was created by counting rows (should equal len(df))
pd.read_sql("SELECT COUNT(*) AS row_count FROM dbo.customer", engine)
```
